```python
import numpy as np
```

```python
im = complex(0.0, 1.0)
xi = 0.5 * (np.sqrt(3) - 1) * (1 + im)
opnorm2 = (1 + np.abs(xi)**2)**2
xi_1 = (1 - np.abs(xi)**2) * (xi + np.conj(xi)) / opnorm2
xi_2 = (np.conj(xi) - xi) * (xi + np.conj(xi)) / opnorm2
xi_3 = (1 - np.abs(xi)**2) * (np.conj(xi) - xi) / opnorm2

def Q(a,b):
    return 1 + (-1)**(( (a >> 1) & 1) + (b & 1)) * xi_1 + (-1)**((a & 1) + ( (a >> 1) & 1) + ( (b >> 1) & 1)) * xi_2 + (-1)**((a & 1) + (b & 1) + ( (b >> 1) & 1)) * xi_3
```

```python
for a in range(0,4):
    for b in range(0,4):
        print("Q( ", (a >> 1) & 1, a&1, ", ", (b >> 1) & 1, b&1, ")")
        print(Q(a,b))
```
