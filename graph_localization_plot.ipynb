```python
%matplotlib ipympl
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.colors import LightSource
import scipy as sp
```

```python
from matplotlib import rc
rc('font',**{'family':'serif','serif':['Times']})
rc('text', usetex=True)
```

```python
n_qubits = 8
filename = "loc_q"+str(n_qubits)+".txt"
symQloc, CAloc = np.genfromtxt("C:/dev/StabilizerStates/data/graphs/"+filename, delimiter=",", dtype = float, usecols=(0,1), unpack = True)
```

```python
fig = plt.figure(6754)
ax = fig.subplots()
ax.set_title(r'$C_A$ localization vs $Q_{mnk}$ localization')
ax.set_xlabel(r'$\sum Q_{mnk}^2$', fontsize=12)
ax.set_ylabel(r'$\sum C_A^2$', fontsize=12)
ax.set_xlim(0,1200)
ax.set_ylim(0,14000)
ax.grid()
ax.scatter(symQloc, CAloc, c='k', s=1)
```
