```python
# Calculates the Q function of the disjoint graph (the 4 qubits graph with connections (1,2), (3,4)) from a direct theoretical calculation
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.colors import LightSource
```

```python
im = complex(0.0, 1.0)
xi = 0.5 * (np.sqrt(3) - 1) * (1 + im)
opxi2 = 1 + np.abs(xi)**2
xi_norm = (1 - np.abs(xi)**2) / opxi2
xi_sum = (xi + np.conj(xi)) / opxi2
xi_min = (np.conj(xi) - xi) / opxi2
```

```python
def sign(a):
    return 1 - 2 * (a & 1)
```

```python
N = 2
qubitstate_size = 2**N
Qfunc = np.zeros((qubitstate_size, qubitstate_size), np.complex128)
for a in range(qubitstate_size):
    a1 = a & 1
    a2 = (a >> 1) & 1
    for b in range(qubitstate_size):
        b1 = b & 1
        b2 = (b >> 1) & 1
        Qfunc[a, b] = 0.25 * (1 + sign(a1 + b2) * xi_norm * xi_sum + sign(a2 + b1) * xi_norm * xi_sum - sign(a1 + a2 + b1 + b2) * xi_min**2)
```

```python
_x = np.arange(qubitstate_size)
_y = np.arange(qubitstate_size)
_xx, _yy = np.meshgrid(_x, _y)
x, y = _xx.ravel(), _yy.ravel()
width = depth = 1
```

```python
light = LightSource(0,20)
fig = plt.figure(figsize=(4,3),dpi=300)
ax = fig.add_subplot(projection='3d')
ax.view_init(elev=25,azim=50)
ax.bar3d(x,y,0,width,depth,Qfunc.flatten().real,lightsource=light)
ax.set_xlabel(r'$\beta$')
ax.set_ylabel(r'$\alpha$')
```

```python
N = 4
qubitstate_size = 2**N
Qfunc = np.zeros((qubitstate_size, qubitstate_size), np.complex128)
for a in range(qubitstate_size):
    a1 = a & 1
    a2 = (a >> 1) & 1
    a3 = (a >> 2) & 1
    a4 = (a >> 3) & 1
    for b in range(qubitstate_size):
        b1 = b & 1
        b2 = (b >> 1) & 1
        b3 = (b >> 2) & 1
        b4 = (b >> 3) & 1
        Qfunc[a, b] = 0.25 * (1 + sign(a1 + b2) * xi_norm * xi_sum + sign(a2 + b1) * xi_norm * xi_sum - sign(a1 + a2 + b1 + b2) * xi_min**2) * 0.25 * (1 + sign(a3 + b4) * xi_norm * xi_sum + sign(a4 + b3) * xi_norm * xi_sum - sign(a3 + a4 + b3 + b4) * xi_min**2)
```

```python
_x = np.arange(qubitstate_size)
_y = np.arange(qubitstate_size)
_xx, _yy = np.meshgrid(_x, _y)
x, y = _xx.ravel(), _yy.ravel()
width = depth = 1
```

```python
light = LightSource(0,20)
fig = plt.figure(figsize=(4,3),dpi=300)
ax = fig.add_subplot(projection='3d')
ax.view_init(elev=25,azim=50)
ax.bar3d(x,y,0,width,depth,Qfunc.flatten().real,lightsource=light)
ax.set_xlabel(r'$\beta$')
ax.set_ylabel(r'$\alpha$')
```
