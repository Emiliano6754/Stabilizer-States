```python
import numpy as np
from matplotlib import pyplot as plt
import math
```

```python
N = 101
mu = 50
sigma2 = 40

norm = (1.0) / (2 ** N)
x = np.arange(0, N+1)

def K1(x):
    return N - 2*x
def K2(x):
    return 2 * (x**2) - 2 * N * x + (1.0 / 2) * N * (N - 1)
def K3(x):
    return - (4.0 / 3) * (x**3) + 2 * N * (x**2) - (N*N - N + (2.0)/3) * x + (1.0/6)*N*(N-1)*(N-2)
g0 = 1
g1 = float(N - 2*mu)/N
g2 = (4.0 / (N*(N-1))) * (sigma2 + mu * (mu - N)) + 1
g3 = (-24*(mu**3) + 64*N*(mu**2) + 2*N*(3 - 8*N + 5*(N**2) + 32*sigma2) - 2*mu*(6 - 9*N + 23*(N**2) + 36*sigma2)) / (3 * N * (N - 1) * (N - 2))
binomials = np.zeros(x.shape, float)
for n in x:
    binomials[n] = math.comb(N, n)
k_gaussian = np.zeros(x.shape, float)
k_gaussian = norm * binomials * (g0 + g1 * K1(x) + g2 * K2(x) + g3 * K3(x))
e_gaussian = (1.0 / np.sqrt(2 * np.pi * sigma2)) * np.exp(- (1.0/(2*sigma2)) * ((x-mu)**2))
```

```python
fig = plt.figure(6754)
ax = fig.subplots()
ax.set_xlim(0,N)
# ax.set_ylim(0,1)
ax.bar(x, k_gaussian, 1, 0, align='edge')
ax.bar(x, e_gaussian, 1, 0, align='edge')
```
